```python

import os
import pandas as pd

DATA_PATH = "/kaggle/working/eye_dataset"
SPLIT_PATH = os.path.join(DATA_PATH, "splits")

# Show all split files
print("Available files:")
print(os.listdir(SPLIT_PATH))

# Inspect each CSV
for filename in sorted(os.listdir(SPLIT_PATH)):
    if filename.endswith(".csv"):
        split_df = pd.read_csv(
            os.path.join(SPLIT_PATH, filename)
        )

        print(f"\n--- {filename} ---")
        print("Total rows:", len(split_df))
        print("Columns:", split_df.columns.tolist())

        display(split_df.head(5))
```
